```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("customer_shopping_behavior.csv")
```

```python
# Print top 5 rows:
df.head()
```

```python
# let see the information about dataset:
df.info()
```

```python
# statistical summary:
df.describe()
```

```python
# check the data has null valus or not:
df.isnull().sum()
```

```python
# fill/replace the null value with it's category median value:
df['Review Rating'] = df.groupby(by='Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
# make our data columns to snake_case so other tools used it efficiently without gives error:
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a column age_group:
labels = ['Young Adult','Adult','Mid-Aged','Senior']
df['age_group'] = pd.qcut(x=df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].sample(10)
```

```python
# check the unique value in a column:
df['frequency_of_purchases'].unique()
```

```python
# create a column purchase_frequency_days:
frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Annually' : 365,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Monthly' : 30,
    'Every 3 Months' : 30
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].sample(10)
```

```python
df[['discount_applied','promo_code_used']]
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
# Both columns are same so delete 'promo_code_used':
df.drop('promo_code_used',axis=1,inplace=True)
```
